# Notebook 07: Live Prediction Engine Demo

### 1. Objective
Demonstrate the end-to-end V3.1 inference pipeline on simulated query scenarios, showing probability rankings and explainability checks.

```python
import os
import sys

# Inject backend modules to path
sys.path.insert(0, os.path.abspath('..'))
from backend.app.ml.v3.predictor import v3_predictor

# Initialize predictor
v3_predictor.load()
print("Prediction Engine successfully initialized!")
```

```python
# Case 1: High Rainfall Climate Query (Rice-like profile)
query_1 = {"N": 80, "P": 40, "K": 40, "temperature": 24.0, "humidity": 82.0, "ph": 6.2, "rainfall": 250.0}
res_1 = v3_predictor.predict(query_1)
print("=== High Rainfall Query Recommendation ===")
for item in res_1['top_recommendations']:
    print(f"Rank {item['rank']}: {item['crop']} | Prob: {item['probability']:.4f}")
```

```python
# Case 2: Extreme Out-of-Distribution Query
query_2 = {"N": 10, "P": 10, "K": 10, "temperature": 5.0, "humidity": 95.0, "ph": 2.0, "rainfall": 10.0}
res_2 = v3_predictor.predict(query_2)
print("=== Out of Distribution Query Result ===")
print(f"Status: {res_2['status']}")
print(f"Message: {res_2['message']}")
```
