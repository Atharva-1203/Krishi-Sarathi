# Notebook 03: Data Cleaning and Forensics

### 1. Objective
Verify data cleanliness, perform missing value analysis, check for duplicates, and distinguish agricultural extreme outliers from invalid data entries.

```python
import pandas as pd
df = pd.read_csv(r'../ml/datasets/v3/dataset.csv')

# Missing value and duplicate checks
nulls = df.isnull().sum().sum()
dupes = df.duplicated().sum()
print(f"Total Missing Values: {nulls}")
print(f"Total Duplicate Rows: {dupes}")
```

```python
# Outlier Analysis - Standard Deviation Method
for col in ['temperature', 'humidity', 'ph', 'rainfall']:
    mean = df[col].mean()
    std = df[col].std()
    outliers = df[(df[col] < mean - 3 * std) | (df[col] > mean + 3 * std)]
    print(f"{col} - Outliers detected beyond 3-sigma: {len(outliers)}")
```

### 2. Interpretation & Judge QA
- **Statistical Outlier vs. Invalid Data**: Outliers in parameters like rainfall (e.g. extremely high rainfall during monsoons) represent real agricultural limits. They are kept because they are agronomically valid. Invalid data (e.g. negative pH) would be rejected.
- **What may a judge ask?** "How do you handle outliers?"
- **Answer:** "We do not clip or drop outliers, as high rainfall or low pH are critical markers for classifying drought-resistant or acid-tolerant crops. Instead, we use min-max scaling to handle bounds gracefully."
