# Notebook 04: Agronomic Feature Analysis

### 1. Objective
Examine feature distributions relative to specific crop classes to map their physiological requirements.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r'../ml/datasets/v3/dataset.csv')

# Crop-wise pH Boxplot for selected crops
selected_crops = ['rice', 'maize', 'chickpea', 'kidneybeans', 'pomegranate', 'mango']
df_sub = df[df['label'].isin(selected_crops)]

plt.figure(figsize=(10, 5))
sns.boxplot(x='label', y='ph', data=df_sub, palette='viridis')
plt.title("pH Ranges for Selected Crop Types")
plt.show()
```

### 2. Interpretation & Judge QA
- **Feature Significance**: pH requirements vary widely. Crops like pomegranate prefer slightly acidic to neutral soils, whereas chickpea is highly robust. This shows how pH act as a strong classifier boundary.
- **What may a judge ask?** "Which parameter has the highest influence on prediction?"
- **Answer:** "Typically, rainfall and humidity have high feature importance coefficients because they act as strict gates for major crop classes (e.g., rice requires high water, whereas chickpea thrives in dry conditions)."
