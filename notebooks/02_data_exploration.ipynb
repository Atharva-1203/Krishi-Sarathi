# Notebook 02: Exploratory Data Analysis (EDA)

### 1. Objective
Examine class distributions, parameter ranges, and correlation matrices to understand the underlying statistical relationships.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r'../ml/datasets/v3/dataset.csv')
sns.set_theme(style="whitegrid")

# Correlation Matrix Heatmap
plt.figure(figsize=(8, 6))
numeric_df = df.drop(columns=['label'])
sns.heatmap(numeric_df.corr(), annot=True, cmap="crest", fmt=".2f")
plt.title("Agronomic Feature Correlation Matrix")
plt.show()
```

```python
# Visualize Distribution of Nitrogen and pH
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(df['N'], kde=True, ax=axes[0], color='emerald')
axes[0].set_title("Nitrogen (N) Distribution")

sns.histplot(df['ph'], kde=True, ax=axes[1], color='teal')
axes[1].set_title("pH Distribution")
plt.show()
```
