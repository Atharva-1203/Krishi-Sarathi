# Notebook 09: Maharashtra Agricultural Analytics Layer

### 1. Objective
Explore the 7.8-lakh records Maharashtra governmental soil database to showcase geographical and regional soil trends.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load processed soil database
soil_db_path = r'../datasets/processed/soil_health/soil_health_database.csv'
soil_df = pd.read_csv(soil_db_path)
print(f"Maharashtra Government Soil Database successfully loaded!")
print(f"Total Records: {soil_df.shape[0]}")
print(f"Columns: {list(soil_df.columns)}")
```

```python
# Group and visualize Nitrogen levels by District
dist_avgs = soil_df.groupby('district')['N'].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10, 5))
dist_avgs.plot(kind='bar', color='olive')
plt.title("Average Soil Nitrogen (N) - Top 10 Districts in Maharashtra")
plt.ylabel("Nitrogen (mg/kg)")
plt.show()
```

### 2. Architectural Concern Separation
- **Decoupled Architecture**: This database feeds the interactive map UI in the Next.js app to display district chemistry. It is **never** used inside the ML model predictor, ensuring that regional defaults do not leak into individual farm recommendations.
