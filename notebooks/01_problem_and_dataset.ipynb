### 1. Objective & Problem Statement
Krishi Sarathi is a data-driven decision support system designed to assist farmers by recommending the most suitable crops for their specific soil chemistry and micro-climatic conditions. Traditional farming methods often rely on generational habits or regional defaults, leading to suboptimal yields, soil degradation, and high financial risks. Supervised Machine Learning addresses this by modeling non-linear crop-suitability thresholds.

### 2. General Prediction vs. Maharashtra Map
